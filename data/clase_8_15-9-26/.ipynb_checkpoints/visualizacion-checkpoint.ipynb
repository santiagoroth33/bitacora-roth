```python
municipios = gpd.read_file(r"C:\Users\esper\anaconda_projects\datos proyecto ciencia de datos\municipios\sig_municipios.shp")
educacion = pd.read_csv(r"C:\Users\esper\anaconda_projects\datos proyecto ciencia de datos\educación\uptu_educacion.csv", sep=";")
policlinicas = pd.read_csv(r"C:\Users\esper\anaconda_projects\datos proyecto ciencia de datos\salud\ubicacion-de-las-policlinicas-de-la-intendencia-de-montevideo.csv", sep=",")
salud = pd.read_csv(r"C:\Users\esper\anaconda_projects\datos proyecto ciencia de datos\salud\uptu_salud.csv", sep=";")
prox_paradas = gpd.read_file(r"C:\Users\esper\anaconda_projects\datos proyecto ciencia de datos\transporte\proximidadparadas.gpkg")
paradas = gpd.read_file(r"C:\Users\esper\anaconda_projects\datos proyecto ciencia de datos\transporte\v_uptu_paradas.shp")
recorrido_omnibus = gpd.read_file(r"C:\Users\esper\anaconda_projects\datos proyecto ciencia de datos\transporte\v_uptu_lsv.shp")
```

```python
#PASAR TODO A TARGET_CRS PARA TENER TODOS LOS DATOS EN EL MISMO MARCO DE COORDENADAS PARA USAR FOLIUM
TARGET_CRS = "EPSG:4326"
gdf_educacion = gpd.GeoDataFrame(
    educacion, 
    geometry=gpd.points_from_xy(educacion['x'], educacion['y']), 
    crs="EPSG:32721"
).to_crs(TARGET_CRS)

gdf_salud = gpd.GeoDataFrame(
    salud, 
    geometry=gpd.points_from_xy(salud['x'], salud['y']), 
    crs="EPSG:32721"
).to_crs(TARGET_CRS)

gdf_policlinicas = gpd.GeoDataFrame(
    policlinicas, 
    geometry=gpd.points_from_xy(policlinicas['Longitud'], policlinicas['Latitud']), 
    crs="EPSG:4326"
)

municipios = municipios.to_crs(TARGET_CRS)         
paradas = paradas.to_crs(TARGET_CRS)
recorrido_omnibus = recorrido_omnibus.to_crs(TARGET_CRS)
prox_paradas = prox_paradas.to_crs(TARGET_CRS)
```
